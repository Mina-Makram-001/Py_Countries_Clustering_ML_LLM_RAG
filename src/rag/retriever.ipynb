```python
from langchain_chroma import Chroma
from langchain_community.embeddings import FastEmbedEmbeddings

# 1. Load the existing vector store from disk
embedding_model = FastEmbedEmbeddings()

vectorstore = Chroma(
    persist_directory="../vectorstore_db",
    embedding_function=embedding_model
)

# 2. Create the retriever interface (top 4 relevant chunks)
retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 4}
)

# 3. Test Retrieval
query = "What is the economic profile of Egypt and what cluster does it belong to?"
results = retriever.invoke(query)

# Print retrieved chunks
for i, doc in enumerate(results, 1):
    print(f"--- Result {i} ---")
    print(f"Metadata: {doc.metadata}")
    print(f"Content:\n{doc.page_content[:300]}...\n")
```
