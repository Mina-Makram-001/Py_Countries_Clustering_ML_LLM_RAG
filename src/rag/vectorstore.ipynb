```python
import json
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
from langchain_community.embeddings import FastEmbedEmbeddings  # or your preferred embedding model

# 1. Load the generated documents
with open("../../docs/all_rag_documents.json", "r", encoding="utf-8") as f:
    raw_docs = json.load(f)

# Convert dict entries back into LangChain Document objects
documents = [
    Document(page_content=doc["page_content"], metadata=doc["metadata"])
    for doc in raw_docs
]

# 2. Chunking & Text Splitting
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150
)
chunked_docs = text_splitter.split_documents(documents)

print(f"Original documents: {len(documents)}")
print(f"Total chunked documents: {len(chunked_docs)}")

# 3. Embed & Save to ChromaDB
embedding_model = FastEmbedEmbeddings()

vectorstore = Chroma.from_documents(
    documents=chunked_docs,
    embedding=embedding_model,
    persist_directory="../vectorstore_db"
)

print("ChromaDB vector store successfully created and persisted!")
```

```python
query = "What countries are in Cluster 2?"

results = vectorstore.similarity_search_with_score(query, k=5)

for i, (doc, score) in enumerate(results):
    print(f"\n{'='*60}")
    print(f"RESULT {i+1}")
    print(f"SCORE: {score}")
    print(f"METADATA: {doc.metadata}")
    print(doc.page_content[:2000])
```
